```python

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["G"],
    "F": [],
    "G": []
}

def dls(graph, node, goal, limit, path):
    if node == goal:
        return "FOUND", path

    children = [
        n for n in graph[node]
        if n not in path
    ]

    if limit == 0:
        if children:
            return "CUTOFF", None
        return "FAILURE", None

    cutoff_found = False

    for child in children:
        status, result_path = dls(
            graph, child, goal,
            limit - 1, path + [child]
        )

        if status == "FOUND":
            return status, result_path

        if status == "CUTOFF":
            cutoff_found = True

    if cutoff_found:
        return "CUTOFF", None

    return "FAILURE", None

def iddfs(graph, start, goal, max_depth):
    attempts = []

    for limit in range(max_depth + 1):
        status, path = dls(
            graph, start, goal, limit, [start]
        )
        attempts.append((limit, status))

        if status == "FOUND":
            return status, path, attempts

    return "CUTOFF", None, attempts

print("DLS RESULTS")

for limit in [1, 2, 3, 4]:
    status, path = dls(
        graph, "A", "G", limit, ["A"]
    )
    print(f"Limit {limit}: {status}, Path: {path}")

print("\nIDDFS RESULTS")

for maximum in [1, 2, 3, 4]:
    status, path, attempts = iddfs(
        graph, "A", "G", maximum
    )
    print(
        f"Maximum {maximum}: "
        f"{status}, Path: {path}, Attempts: {attempts}"
    )
```
