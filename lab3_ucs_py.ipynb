```python

import heapq
from collections import deque

graph = {
    "A": [("B", 1), ("C", 10)],
    "B": [("D", 4), ("E", 1)],
    "C": [("F", 2), ("G", 2)],
    "D": [], "E": [("G", 1)], "F": [], "G": []
}

def bfs():
    q, seen = deque([("A", ["A"])]), {"A"}
    while q:
        node, path = q.popleft()
        if node == "G":
            return path
        for nxt, _ in graph[node]:
            if nxt not in seen:
                seen.add(nxt)
                q.append((nxt, path + [nxt]))

def ucs():
    q, best, order = [(0, "A", ["A"])], {"A": 0}, []
    while q:
        cost, node, path = heapq.heappop(q)
        if cost != best.get(node):
            continue
        order.append((node, cost))
        if node == "G":
            return path, cost, order
        for nxt, w in graph[node]:
            new_cost = cost + w
            if new_cost < best.get(nxt, float("inf")):
                best[nxt] = new_cost
                heapq.heappush(q, (new_cost, nxt, path + [nxt]))

def cost(path):
    return sum(next(w for n, w in graph[a] if n == b)
               for a, b in zip(path, path[1:]))

bpath = bfs()
upath, ucost, order = ucs()

print("BFS Path:", bpath, "Cost:", cost(bpath))
print("UCS Path:", upath)
print("UCS Processing Order:", order)
print("UCS Returned Cost:", ucost)
print("Verified Cost:", cost(upath))

```
