```python
class ClassroomAgent:
    def __init__(self):
        self.previous_mode = None

    def get_target_mode(self, temp, occupied):
        # Occupancy has priority over temperature
        if not occupied:
            return "ECO"

        if temp > 26:
            return "COOL"

        if temp < 20:
            return "WARM"

        return "IDLE"

    def act(self, temp, occupied):
        target_mode = self.get_target_mode(temp, occupied)

        # Send command only if target mode is different
        # from the previous mode
        if target_mode != self.previous_mode:
            command = "SEND"
        else:
            command = "NO COMMAND"

        # Remember the current target mode
        self.previous_mode = target_mode

        return target_mode, command


agent = ClassroomAgent()

percepts = [
    (29, True),
    (29, True),
    (20, True),
    (26, True),
    (19, True),
    (29, False)
]

for step, percept in enumerate(percepts, 1):
    temp, occupied = percept

    previous_mode = agent.previous_mode
    target_mode, command = agent.act(temp, occupied)

    print(
        step,
        "Previous:", previous_mode,
        "Percept:", percept,
        "Target:", target_mode,
        "Command:", command
    )
```
